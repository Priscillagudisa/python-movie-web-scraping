```python
import requests
import sqlite3
import pandas as pd
from bs4 import BeautifulSoup


url = "https://web.archive.org/web/20230902185655/https://en.everybodywiki.com/100_Most_Highly-Ranked_Films"

response = requests.get(url, timeout=30)

print("Status code:", response.status_code)
print("HTML length:", len(response.text))
print(response.text[:500])
```

```python
data = BeautifulSoup(response.text, "html.parser")

tables = data.find_all("table")
tbodies = data.find_all("tbody")

print("Number of tables:", len(tables))
print("Number of tbody elements:", len(tbodies))
```

```python
tables = data.find_all("table")

if len(tables) == 0:
    print("No HTML tables found. Check the webpage response.")
else:
    rows = tables[0].find_all("tr")
    print("Rows found:", len(rows))
```

```python
movies = []

for row in rows:
    col = row.find_all("td")

    if len(col) >= 3:
        data_dict = {
            "Average Rank": col[0].get_text(strip=True),
            "Film": col[1].get_text(strip=True),
            "Year": col[2].get_text(strip=True)
        }

        movies.append(data_dict)

    if len(movies) == 50:
        break

df = pd.DataFrame(movies)

print(df)
```
